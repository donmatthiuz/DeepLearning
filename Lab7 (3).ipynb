###  Tiny Llama

````python
model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"  # or "Qwen/Qwen2.5-0.5B-Instruct"

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

use_4bit = True  # set False if you prefer full precision

dtype = torch.bfloat16 if torch.cuda.is_available() and torch.cuda.is_bf16_supported() else torch.float16

tokenizer = AutoTokenizer.from_pretrained(model_id)

if use_4bit:
    # Memory efficient loading for small VRAM GPUs (needs bitsandbytes)
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        device_map="auto",
        load_in_4bit=True
    )
else:
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=dtype,
        device_map="auto"
    )

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    # return_full_text=False lets us see just the new tokens for some models
    return_full_text=False,
)
````

````python
prompt = """Explain me whats a transformer, assuming im a 5 years old kid, transformes in deep learning"""
outputs = generator(
    prompt,
    max_new_tokens=200,
    temperature=0.2
)
print(outputs[0]['generated_text'])
````

###  Doctor-Shotgun

````python
modelo_id = "Doctor-Shotgun/TinyLlama-1.1B-32k-Instruct"

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

usar_4bit = True
tipo_dato = torch.bfloat16 if torch.cuda.is_available() and torch.cuda.is_bf16_supported() else torch.float16

tokenizador = AutoTokenizer.from_pretrained(modelo_id)

if usar_4bit:
    modelo = AutoModelForCausalLM.from_pretrained(
        modelo_id,
        device_map="auto",
        load_in_4bit=True
    )
else:
    modelo = AutoModelForCausalLM.from_pretrained(
        modelo_id,
        torch_dtype=tipo_dato,
        device_map="auto"
    )

creador_texto = pipeline(
    "text-generation",
    model=modelo,
    tokenizer=tokenizador,
    return_full_text=False,
)
````

````python
prompt = "Explain me whats a transformer, assuming im a 5 years old kid, transformes in deep learning"

outputs = creador_texto(
    prompt,
    max_new_tokens=200,
    temperature=0.2,
    top_p=0.9
)

print(outputs[0]["generated_text"])
````

## Ficha Tecnica




### **1. Módulos Utilizados**

| Módulo            | Descripción                                                                                                             | Función principal en el laboratorio                                                                                              |
| :---------------- | :---------------------------------------------------------------------------------------------------------------------- | :------------------------------------------------------------------------------------------------------------------------------- |
| **torch**         | Framework de *deep learning* de PyTorch. Permite manejar tensores, tipos de datos (float16, bfloat16), y el uso de GPU. | Define el tipo de dato (`torch.bfloat16` o `torch.float16`) y gestiona el uso de CUDA.                                           |
| **transformers**  | Librería de Hugging Face que contiene modelos preentrenados y herramientas para NLP.                                    | Se usa para cargar modelos (`AutoModelForCausalLM`), tokenizadores (`AutoTokenizer`) y crear *pipelines* de generación de texto. |
| **bitsandbytes**  | Librería para carga eficiente de modelos en baja precisión (4 bits o 8 bits).                                           | Reduce el consumo de VRAM usando `load_in_4bit=True`. Es esencial en GPUs con poca memoria.                                      |
| **accelerate**    | Gestiona la distribución del modelo en múltiples dispositivos y optimiza el rendimiento.                                | Permite que `device_map="auto"` detecte automáticamente el mejor dispositivo para el modelo.                                     |
| **sentencepiece** | Tokenizador subword usado por muchos modelos de Hugging Face.                                                           | Permite dividir texto en *tokens* adecuados para el modelo.                                                                      |
| **openai**        | Cliente oficial de la API de OpenAI.                                                                                    | Permite conectarse al modelo **GPT-4** vía la API para generar respuestas.                                                       |

---

### **2. Funciones Clave**

| Función                                                 | Módulo         | Descripción                                                                      | Parámetros importantes                                         |
| :------------------------------------------------------ | :------------- | :------------------------------------------------------------------------------- | :------------------------------------------------------------- |
| **`AutoTokenizer.from_pretrained(model_id)`**           | `transformers` | Carga el tokenizador correspondiente al modelo especificado.                     | `model_id`: nombre del modelo en Hugging Face.                 |
| **`AutoModelForCausalLM.from_pretrained(model_id, …)`** | `transformers` | Carga el modelo de lenguaje preentrenado para tareas de generación causal.       | `device_map="auto"`, `load_in_4bit=True`, `torch_dtype=dtype`. |
| **`pipeline("text-generation", …)`**                    | `transformers` | Crea una interfaz lista para generar texto con el modelo y tokenizador.          | `model`, `tokenizer`, `return_full_text=False`.                |
| **`OpenAI(api_key=...)`**                               | `openai`       | Inicializa el cliente con una clave API de OpenAI.                               | `api_key`: clave personal o del proyecto.                      |
| **`client.chat.completions.create(...)`**               | `openai`       | Envía un *prompt* al modelo GPT (por ejemplo, GPT-4) para recibir una respuesta. | `model`, `messages`, `temperature`, `max_tokens`.              |

---

### **4. Parámetros Técnicos Relevantes**

| Parámetro                     | Valor                              | Función                                                                   |
| :---------------------------- | :--------------------------------- | :------------------------------------------------------------------------ |
| `use_4bit` / `usar_4bit`      | `True`                             | Activa la cuantización a 4 bits para ahorrar memoria en GPU.              |
| `dtype` / `tipo_dato`         | `torch.bfloat16` o `torch.float16` | Define la precisión del modelo (baja precisión para menor consumo).       |
| `device_map`                  | `"auto"`                           | Asigna automáticamente CPU/GPU según disponibilidad.                      |
| `return_full_text`            | `False`                            | Muestra solo el texto generado, sin repetir el prompt original.           |
| `temperature` (en OpenAI API) | `0.2`                              | Controla la aleatoriedad de la respuesta (valores bajos = más precisión). |
| `max_tokens`                  | `200`                              | Limita la longitud de la salida del modelo.                               |
